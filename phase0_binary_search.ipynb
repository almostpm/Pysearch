```python
def binary_search(nums, target):
    # Initialize the lower bound pointer to the array start
    low = 0
    # Set the higher bound pointer to the final index
    high = len(nums) - 1
    
    # Continue searching until the boundary pointers cross each other
    while low <= high:
        # Calculate the middle index using floor integer division
        mid = (low + high) // 2
        # Retrieve the value stored at the middle index position
        guess = nums[mid]
        
        # Check if the middle value matches the target item
        if guess == target:
            # Return the correct matching index position immediately
            return mid
        # Check if the middle value is too low
        elif guess < target:
            # Shift the lower bound pointer past the mid index
            low = mid + 1
        # Execute if the middle value is too high
        else:
            # Shift the higher bound pointer below the mid index
            high = mid - 1
    
    # Return None if the target value is not found
    return None

# Tests
# Verify target inside a basic odd-length list array
print(binary_search([1, 3, 5, 7, 9], 5))    # 2
# Verify target located at the absolute lowest index boundary
print(binary_search([1, 3, 5, 7, 9], 1))    # 0
# Verify target located at the absolute highest index boundary
print(binary_search([1, 3, 5, 7, 9], 9))    # 4
# Verify behavior when search item is missing from sequence
print(binary_search([1, 3, 5, 7, 9], 4))    # None
# Verify behavior when evaluating an entirely empty list input
print(binary_search([], 5))                 # None
# Verify single element array execution handles target hits cleanly
print(binary_search([5], 5))                # 0
```
