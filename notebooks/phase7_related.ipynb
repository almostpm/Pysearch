```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("PROJECT_ROOT:", PROJECT_ROOT)
```

```python
from src.related import build_graph, bfs, related

test_docs = [
    {"id": 0, "name": "d0.txt", "content": "python python search"},
    {"id": 1, "name": "d1.txt", "content": "search engine"},
    {"id": 2, "name": "d2.txt", "content": "python hash table"},
    {"id": 3, "name": "d3.txt", "content": ""},
]

graph = build_graph(test_docs)
print("Graph built with", len(graph), "nodes")
```

```python
assert isinstance(graph, dict)
assert all(isinstance(v, set) for v in graph.values())
print("✅ Test 1 passed")
```

```python
neighbors = graph.get("python", set())
assert "search" in neighbors or "hash" in neighbors
print("✅ Test 2 passed")
```

```python
distances = bfs(graph, "python", 2)
assert distances.get("python") == 0
print("✅ Test 3 passed")
```

```python
distances = bfs(graph, "xyz", 2)
assert distances == {}
print("✅ Test 4 passed")
```

```python
result = related("python", graph)
assert isinstance(result, list)
assert "python" not in result
print("✅ Test 5 passed")
```

```python
result = related("xyz", graph)
assert result == []
print("✅ Test 6 passed")
```

```python
result = related("python", graph, max_results=1)
assert len(result) <= 1
print("✅ Test 7 passed")
```

```python
distances = bfs(graph, "python", 2)
result = related("python", graph)
result_distances = [distances[w] for w in result]
assert result_distances == sorted(result_distances)
print("✅ Test 8 passed")
```
