```python
import os  # import OS module
import sys  # import sys module
if os.path.basename(os.getcwd()) == 'notebooks':  # check if in notebooks dir
    os.chdir('..')  # move up one directory
sys.path.append(os.getcwd())  # add root to sys path
assert os.path.exists('src'), 'src/ folder not found!'  # verify src exists
```

```python
import math  # import math for log
from src.ranker import compute_tf, compute_idf, rank  # import ranker functions

test_docs = [  # define test fixture
    {"id": 0, "name": "d0.txt", "content": "python python search"},  # doc 0
    {"id": 1, "name": "d1.txt", "content": "search engine"},  # doc 1
    {"id": 2, "name": "d2.txt", "content": "python hash table"},  # doc 2
    {"id": 3, "name": "d3.txt", "content": ""}  # doc 3 empty
]  # end of fixture list
```

```python
assert math.isclose(compute_tf("python", "python python search"), 2/3)  # TF check
print("✅ Test 1 passed")  # success message
```

```python
assert compute_tf("xyz", "python python search") == 0.0  # TF unknown term
print("✅ Test 2 passed")  # success message
```

```python
assert compute_tf("python", "") == 0.0  # TF empty doc
print("✅ Test 3 passed")  # success message
```

```python
assert math.isclose(compute_idf("python", test_docs), math.log(4/2))  # IDF check
print("✅ Test 4 passed")  # success message
```

```python
assert math.isclose(compute_idf("engine", test_docs), math.log(4/1))  # IDF check
print("✅ Test 5 passed")  # success message
```

```python
assert compute_idf("xyz", test_docs) == 0.0  # IDF unknown term
print("✅ Test 6 passed")  # success message
```

```python
assert compute_idf("python", []) == 0.0  # IDF empty docs
print("✅ Test 7 passed")  # success message
```

```python
result = rank([0, 2], "python", test_docs)  # run rank function
assert result[0][0] == 0  # doc 0 has higher score than doc 2
print("✅ Test 8 passed")  # success message
```

```python
tie_docs = [{"id": 5, "content": "tie"}, {"id": 2, "content": "tie"}]  # exact same content
result = rank([2, 5], "tie", tie_docs)  # rank them
assert result[0][0] == 2  # lower id wins
assert result[1][0] == 5  # higher id loses
print("✅ Test 9 passed")  # success message
```

```python
assert rank([], "python", test_docs) == []  # empty query -> []
assert rank([0, 1], "", test_docs) == []  # empty ids -> []
try:  # start try block
    rank([99], "python", test_docs)  # unknown id
    raise AssertionError("ValueError not raised for unknown doc_id")  # should fail before here
except ValueError:  # catch ValueError
    pass  # all good
print("✅ Test 10 passed")  # success message
```
