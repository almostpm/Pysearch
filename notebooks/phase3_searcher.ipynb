```python
import os  # import OS module
import sys  # import sys module
if os.path.basename(os.getcwd()) == 'notebooks':  # check if in notebooks dir
    os.chdir('..')  # move up one directory
sys.path.append(os.getcwd())  # add root to sys path
assert os.path.exists('src'), 'src/ folder not found!'  # verify src exists
```

```python
from src.indexer import build_index  # import build_index
from src.searcher import search  # import search

test_docs = [  # define test fixture
    {"id": 0, "name": "d0.txt", "content": "python python search"},  # doc 0
    {"id": 1, "name": "d1.txt", "content": "search engine"},  # doc 1
    {"id": 2, "name": "d2.txt", "content": "python hash table"},  # doc 2
    {"id": 3, "name": "d3.txt", "content": ""}  # doc 3 empty
]
test_index = build_index(test_docs)  # build the index
# Expected: {"python": [0, 2], "search": [0, 1], "engine": [1], "hash": [2], "table": [2]}
assert test_index == {"python": [0, 2], "search": [0, 1], "engine": [1], "hash": [2], "table": [2]}  # verify index
```

```python
assert search("python", test_index) == [0, 2]  # single known word
print("✅ Test 1 passed")  # success message
```

```python
assert search("xyz", test_index) == []  # unknown word
print("✅ Test 2 passed")  # success message
```

```python
assert search("python hash", test_index, "AND") == [2]  # AND multi-word
print("✅ Test 3 passed")  # success message
```

```python
assert search("python xyz", test_index, "AND") == []  # AND with unknown word
print("✅ Test 4 passed")  # success message
```

```python
assert sorted(search("engine hash", test_index, "OR")) == [1, 2]  # OR mode
print("✅ Test 5 passed")  # success message
```

```python
assert sorted(search("python OR hash", test_index)) == [0, 2]  # reserved 'or' keyword overrides AND
print("✅ Test 6 passed")  # success message
```

```python
assert search("", test_index) == []  # empty query
print("✅ Test 7 passed")  # success message
```

```python
assert search("PYTHON", test_index) == [0, 2]  # case-insensitive
print("✅ Test 8 passed")  # success message
```

```python
assert search("search python", test_index, "OR") == [0, 1, 2]  # sorted output naturally
print("✅ Test 9 passed")  # success message
```

```python
try:  # start try block for exception catching
    search("python", test_index, mode="INVALID")  # attempt invalid mode
    raise AssertionError("ValueError was not raised")  # fail if it goes through
except ValueError:  # catch expected error
    pass  # all good
print("✅ Test 10 passed")  # success message
```
