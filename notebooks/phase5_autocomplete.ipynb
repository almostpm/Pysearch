```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("PROJECT_ROOT:", PROJECT_ROOT)
```

```python
from src.autocomplete import build_trie, autocomplete

words = ["python", "pytorch", "pycharm", "search", "engine", "hash", "table"]
trie = build_trie(words)
print("Trie built with", len(words), "words")
```

```python
result = autocomplete("py", trie)
assert result == sorted(result), "Result should be sorted"
assert all(w.startswith("py") for w in result), "All results should start with 'py'"
print("✅ Test 1 passed")
```

```python
result = autocomplete("xyz", trie)
assert result == []
print("✅ Test 2 passed")
```

```python
result = autocomplete("", trie)
assert len(result) <= 10
assert set(result).issubset(set(words))
print("✅ Test 3 passed")
```

```python
result_lower = autocomplete("py", trie)
result_upper = autocomplete("PY", trie)
assert result_lower == result_upper
print("✅ Test 4 passed")
```

```python
result = autocomplete("py", trie, max_results=2)
assert len(result) == 2
print("✅ Test 5 passed")
```

```python
try:
    autocomplete(123, trie)
    raised = False
except TypeError:
    raised = True
assert raised
print("✅ Test 6 passed")
```

```python
empty_trie = build_trie([])
result = autocomplete("py", empty_trie)
assert result == []
print("✅ Test 7 passed")
```
