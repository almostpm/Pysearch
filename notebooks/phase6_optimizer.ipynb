```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / "src").exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("PROJECT_ROOT:", PROJECT_ROOT)
```

```python
from src.optimizer import cached_tokenize, get_stats, measure_search_time, measure_rank_time, comparison_report
from src.indexer import build_index

test_docs = [
    {"id": 0, "name": "d0.txt", "content": "python python search"},
    {"id": 1, "name": "d1.txt", "content": "search engine"},
    {"id": 2, "name": "d2.txt", "content": "python hash table"},
    {"id": 3, "name": "d3.txt", "content": ""},
]

test_index = build_index(test_docs)
print("Index built with", len(test_index), "unique words")
```

```python
result = cached_tokenize("hello world")
assert result == ["hello", "world"]
print("✅ Test 1 passed")
```

```python
result_a = cached_tokenize("same")
result_b = cached_tokenize("same")
assert result_a is result_b
print("✅ Test 2 passed")
```

```python
stats = get_stats(test_docs, test_index)
expected_keys = {"total_docs", "total_tokens", "unique_words", "avg_tokens_per_doc", "index_size_bytes"}
assert set(stats.keys()) == expected_keys
print("✅ Test 3 passed")
```

```python
stats = get_stats([], {})
assert stats["total_docs"] == 0
assert stats["total_tokens"] == 0
assert stats["unique_words"] == 0
assert stats["avg_tokens_per_doc"] == 0.0
assert stats["index_size_bytes"] == 0
print("✅ Test 4 passed")
```

```python
avg_ms = measure_search_time("python", test_index, iterations=10)
assert isinstance(avg_ms, float)
assert avg_ms > 0
print("✅ Test 5 passed")
```

```python
avg_ms = measure_rank_time([0, 2], "python", test_docs, iterations=10)
assert isinstance(avg_ms, float)
assert avg_ms > 0
print("✅ Test 6 passed")
```

```python
report = comparison_report(test_docs, test_index)
assert "naive_ms" in report
assert "cached_ms" in report
assert "speedup" in report
print("✅ Test 7 passed")
```

```python
report = comparison_report(test_docs, test_index)
assert report["speedup"] >= 1.0
print("✅ Test 8 passed")
```
